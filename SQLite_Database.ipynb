# SQL Analysis for IT Helpdesk Ticket Analytics

### SQLite | SQL Queries

## Project Objective

This notebook imports the cleaned IT helpdesk dataset into SQLite and performs SQL queries to answer common business questions.

```python
from google.colab import files

uploaded = files.upload()
```

```python
import pandas as pd
import sqlite3
```

```python
df = pd.read_csv("cleaned_ticket.csv")
```

## Load Cleaned Dataset

```python
df.head()
```

```python
conn = sqlite3.connect("helpdesk.db")
```

```python
df.to_sql(
    "Tickets",
    conn,
    if_exists="replace",
    index=False
)
```

## Query 1: Display the First Five Records

```python
query = """
SELECT *
FROM Tickets
LIMIT 5;
"""

pd.read_sql(query, conn)
```

## Query 2: Count Total Number of Tickets

```python
query = """
SELECT COUNT(*) AS Total_Tickets
FROM Tickets;
"""

pd.read_sql(query, conn)
```

## Query 3: Count Tickets by Department

```python
query = """
SELECT Department,
COUNT(*) AS Total_Tickets

FROM Tickets

GROUP BY Department

ORDER BY Total_Tickets DESC;
"""

pd.read_sql(query, conn)
```

## Query 4: Count Tickets by Priority

```python
query = """
SELECT Priority,
COUNT(*) AS Total

FROM Tickets

GROUP BY Priority

ORDER BY Total DESC;
"""

pd.read_sql(query, conn)
```

## Query 5: Display High-Priority Tickets

```python
query = """
SELECT *
FROM Tickets
WHERE Priority='high';
"""

pd.read_sql(query, conn)
```

## Query 6: Display Low-Priority Tickets

```python
query = """
SELECT *
FROM Tickets
WHERE Priority='low';
"""

pd.read_sql(query, conn)
```

## Query 7: Department-wise Priority Distribution

```python
query = """
SELECT Department,
Priority,
COUNT(*) AS Total

FROM Tickets

GROUP BY Department, Priority

ORDER BY Department;
"""

pd.read_sql(query, conn)
```

## Query 8: Department with the Highest Number of Tickets

```python
query = """
SELECT Department,
COUNT(*) AS Tickets

FROM Tickets

GROUP BY Department

ORDER BY Tickets DESC

LIMIT 1;
"""

pd.read_sql(query, conn)
```

## Query 9: Count Total Departments

```python
query = """
SELECT COUNT(DISTINCT Department) AS Total_Departments
FROM Tickets;
"""

pd.read_sql(query, conn)
```

## Query 10: Priority-wise Ticket Count

```python
query = """
SELECT Priority,
COUNT(*) AS Total

FROM Tickets

GROUP BY Priority;
"""

pd.read_sql(query, conn)
```

```python
department_sql = pd.read_sql("""
SELECT Department,
COUNT(*) AS Total_Tickets

FROM Tickets

GROUP BY Department
ORDER BY Total_Tickets DESC;
""", conn)

department_sql.to_csv("department_sql.csv", index=False)
```

# SQL Summary

The SQL queries confirmed the insights obtained from Python analysis.

- Technical Support received the highest number of tickets.
- Medium-priority tickets were the most common.
